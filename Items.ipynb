```python
import pandas as pd
language = 'es'
dct = {}
for k,v in items.items():
    item_set = get_item_set(v,item_sets)
    dct[k] = {
        "name" : v["name"][language],
        "set" : item_set["name"][language] if item_set else None,
        "level" : v["level"],
        'ankama_id' : v['ankama_id']
    }
    dct[k].update({i:j for i,j in v['type'].items() if i not in dct[k]})
Items = pd.DataFrame.from_dict(dct, orient='index')
Items.to_csv('data/ITEM_NAMES_AND_IDS.csv', index=False)
```

```python
# with open("reports/crit-omni-ranged/2026-09-05_15-44-02/config.json", "r") as file:
with open("workflows/report-sacri-tank_20260922-110152/run_training/config.json", "r") as file:
    config = json.load(file)
    # Convert all string keys into int
    config['character']['exo'] = {int(k): v for k, v in config['character']['exo'].items()}
```

```python
import pandas as pd

dct = {}
for k,v in items.items():
    item_set = get_item_set(v,item_sets)
    dct[k] = {
        "name" : v["name"][config["optimizer"]["language"]],
        "set" : item_set["name"][config["optimizer"]["language"]] if item_set else None,
        "level" : v["level"],
        'ankama_id' : v['ankama_id']
    }
    dct[k].update({i:j for i,j in v['type'].items() if i not in dct[k]})
Items = pd.DataFrame.from_dict(dct, orient='index')
```

```python
char = Character(**config["character"])
opt = Optimizer(char,
                 items,
                   item_sets,
                     **config["optimizer"])
```

```python
import pygad
from scripts import Optimizer
opt = Optimizer()
```

```python
def fitness(ga_instance,chromosome,chromosome_idx):
    # Penalizations
    if len(set(chromosome[-opt.pool_types[-1][1]:])) < opt.pool_types[-1][1]:
        # Some dofus or trophies are duplicated
        return 0
    if len(set(chromosome[2:4])) < 2:
        # Some rings are duplicated
        return 0

    weapon = opt.get_chromosome_weapon(chromosome)
    return opt.get_damage(weapon,chromosome)
```

```python
# Example setup
ga_instance = pygad.GA(
    num_generations=1000,
    num_parents_mating=150,
    sol_per_pop=150,
    num_genes=len(opt.pools),
    fitness_func=fitness,
    gene_type=int,
    gene_space=opt.pools,
    parent_selection_type="tournament",
    crossover_probability=0.8,
    mutation_probability=0.05,
    keep_elitism=2
)
```

```python
ga_instance.run()
```

```python
import numpy as np
solution, solution_fitness, solution_idx = ga_instance.best_solution()
print("Parameters of the best solution : {solution}".format(solution=solution))
print("Fitness value of the best solution = {solution_fitness}".format(solution_fitness=solution_fitness))
```

```python
for k,v in opt.get_totals(solution).items():
    print(f"({k}) {opt.item_descriptions[k]}: {v}")
```
